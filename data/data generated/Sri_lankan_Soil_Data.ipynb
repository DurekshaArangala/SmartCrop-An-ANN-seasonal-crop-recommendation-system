```python
# ================================================================
# SMARTCROP
# SYNTHETIC SRI LANKAN SOIL DATASET
# ================================================================
#
# Generates:
#   5,000 synthetic soil records
#
# Columns:
#   Sample_ID
#   District
#   Latitude
#   Longitude
#   Agro_Ecological_Zone
#   Soil_Order_Classification
#   pH
#   Nitrogen
#   Phosphorus
#   Potassium
#   Organic_Carbon
#   Sand
#   Clay
#   Silt
#
# Output:
#   soil_data.csv
#
# ================================================================


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files

np.random.seed(42)


# ================================================================
# 2. NUMBER OF RECORDS
# ================================================================

N = 5000

print("Number of records to generate:", N)


# ================================================================
# 3. SRI LANKAN DISTRICTS
# ================================================================

districts = {

    "Colombo": {
        "lat": (6.80, 6.98),
        "lon": (79.80, 80.05),
        "zone": "Wet Zone",
        "soil": "Ultisols",
        "ph": 5.8,
        "oc": 1.7,
        "n": 1.20,
        "p": 14,
        "k": 120,
        "sand": 55,
        "clay": 25
    },

    "Gampaha": {
        "lat": (7.00, 7.25),
        "lon": (79.85, 80.15),
        "zone": "Wet Zone",
        "soil": "Ultisols",
        "ph": 5.7,
        "oc": 1.6,
        "n": 1.15,
        "p": 13,
        "k": 115,
        "sand": 58,
        "clay": 23
    },

    "Kalutara": {
        "lat": (6.40, 6.75),
        "lon": (79.95, 80.30),
        "zone": "Wet Zone",
        "soil": "Ultisols",
        "ph": 5.5,
        "oc": 1.8,
        "n": 1.30,
        "p": 15,
        "k": 125,
        "sand": 50,
        "clay": 28
    },

    "Kandy": {
        "lat": (7.20, 7.45),
        "lon": (80.55, 80.75),
        "zone": "Wet Zone",
        "soil": "Ultisols",
        "ph": 5.4,
        "oc": 2.1,
        "n": 1.45,
        "p": 16,
        "k": 135,
        "sand": 48,
        "clay": 32
    },

    "Matale": {
        "lat": (7.40, 7.80),
        "lon": (80.55, 80.85),
        "zone": "Intermediate Zone",
        "soil": "Alfisols",
        "ph": 5.6,
        "oc": 1.9,
        "n": 1.30,
        "p": 14,
        "k": 125,
        "sand": 55,
        "clay": 27
    },

    "Nuwara Eliya": {
        "lat": (6.85, 7.10),
        "lon": (80.65, 80.90),
        "zone": "Up Country Wet Zone",
        "soil": "Inceptisols",
        "ph": 4.8,
        "oc": 3.2,
        "n": 1.70,
        "p": 18,
        "k": 145,
        "sand": 40,
        "clay": 38
    },

    "Galle": {
        "lat": (5.95, 6.20),
        "lon": (80.10, 80.35),
        "zone": "Wet Zone",
        "soil": "Ultisols",
        "ph": 5.3,
        "oc": 2.0,
        "n": 1.35,
        "p": 16,
        "k": 130,
        "sand": 48,
        "clay": 30
    },

    "Matara": {
        "lat": (5.90, 6.15),
        "lon": (80.45, 80.70),
        "zone": "Wet Zone",
        "soil": "Ultisols",
        "ph": 5.4,
        "oc": 1.85,
        "n": 1.25,
        "p": 15,
        "k": 125,
        "sand": 52,
        "clay": 28
    },

    "Hambantota": {
        "lat": (6.00, 6.35),
        "lon": (80.55, 81.10),
        "zone": "Dry Zone",
        "soil": "Alfisols",
        "ph": 6.5,
        "oc": 1.1,
        "n": 0.90,
        "p": 10,
        "k": 110,
        "sand": 68,
        "clay": 17
    },

    "Jaffna": {
        "lat": (9.55, 9.85),
        "lon": (79.70, 80.20),
        "zone": "Dry Zone",
        "soil": "Alfisols",
        "ph": 7.5,
        "oc": 0.9,
        "n": 0.75,
        "p": 12,
        "k": 170,
        "sand": 70,
        "clay": 15
    },

    "Kilinochchi": {
        "lat": (9.30, 9.65),
        "lon": (80.25, 80.65),
        "zone": "Dry Zone",
        "soil": "Alfisols",
        "ph": 6.9,
        "oc": 1.0,
        "n": 0.85,
        "p": 11,
        "k": 155,
        "sand": 65,
        "clay": 18
    },

    "Mannar": {
        "lat": (8.80, 9.20),
        "lon": (79.80, 80.10),
        "zone": "Dry Zone",
        "soil": "Alfisols",
        "ph": 7.2,
        "oc": 0.8,
        "n": 0.70,
        "p": 10,
        "k": 160,
        "sand": 75,
        "clay": 13
    },

    "Mullaitivu": {
        "lat": (9.00, 9.40),
        "lon": (80.45, 81.00),
        "zone": "Dry Zone",
        "soil": "Alfisols",
        "ph": 6.8,
        "oc": 1.0,
        "n": 0.85,
        "p": 12,
        "k": 150,
        "sand": 67,
        "clay": 18
    },

    "Vavuniya": {
        "lat": (8.60, 9.00),
        "lon": (80.30, 80.70),
        "zone": "Dry Zone",
        "soil": "Alfisols",
        "ph": 6.6,
        "oc": 1.2,
        "n": 0.95,
        "p": 13,
        "k": 145,
        "sand": 62,
        "clay": 22
    },

    "Batticaloa": {
        "lat": (7.50, 8.00),
        "lon": (81.00, 81.60),
        "zone": "Dry Zone",
        "soil": "Vertisols",
        "ph": 6.8,
        "oc": 1.3,
        "n": 1.00,
        "p": 14,
        "k": 150,
        "sand": 60,
        "clay": 22
    },

    "Ampara": {
        "lat": (6.90, 7.50),
        "lon": (81.40, 81.90),
        "zone": "Dry Zone",
        "soil": "Vertisols",
        "ph": 6.4,
        "oc": 1.45,
        "n": 1.10,
        "p": 15,
        "k": 145,
        "sand": 58,
        "clay": 25
    },

    "Trincomalee": {
        "lat": (8.40, 8.90),
        "lon": (80.70, 81.50),
        "zone": "Dry Zone",
        "soil": "Alfisols",
        "ph": 6.7,
        "oc": 1.2,
        "n": 0.95,
        "p": 12,
        "k": 140,
        "sand": 65,
        "clay": 20
    },

    "Kurunegala": {
        "lat": (7.30, 8.00),
        "lon": (80.10, 80.70),
        "zone": "Intermediate Zone",
        "soil": "Alfisols",
        "ph": 6.2,
        "oc": 1.3,
        "n": 1.00,
        "p": 12,
        "k": 130,
        "sand": 62,
        "clay": 21
    },

    "Puttalam": {
        "lat": (7.60, 8.30),
        "lon": (79.70, 80.20),
        "zone": "Dry Zone",
        "soil": "Alfisols",
        "ph": 6.7,
        "oc": 0.95,
        "n": 0.80,
        "p": 10,
        "k": 120,
        "sand": 75,
        "clay": 14
    },

    "Anuradhapura": {
        "lat": (7.90, 8.60),
        "lon": (80.30, 80.90),
        "zone": "Dry Zone",
        "soil": "Alfisols",
        "ph": 6.3,
        "oc": 1.0,
        "n": 0.85,
        "p": 11,
        "k": 125,
        "sand": 68,
        "clay": 17
    },

    "Polonnaruwa": {
        "lat": (7.70, 8.20),
        "lon": (80.80, 81.30),
        "zone": "Dry Zone",
        "soil": "Vertisols",
        "ph": 6.4,
        "oc": 1.2,
        "n": 1.00,
        "p": 13,
        "k": 140,
        "sand": 60,
        "clay": 22
    },

    "Badulla": {
        "lat": (6.80, 7.20),
        "lon": (80.85, 81.20),
        "zone": "Up Country Intermediate Zone",
        "soil": "Inceptisols",
        "ph": 5.2,
        "oc": 2.3,
        "n": 1.50,
        "p": 16,
        "k": 140,
        "sand": 45,
        "clay": 34
    },

    "Monaragala": {
        "lat": (6.60, 7.30),
        "lon": (81.00, 81.60),
        "zone": "Intermediate Zone",
        "soil": "Alfisols",
        "ph": 5.9,
        "oc": 1.3,
        "n": 1.00,
        "p": 12,
        "k": 130,
        "sand": 63,
        "clay": 20
    },

    "Ratnapura": {
        "lat": (6.55, 6.90),
        "lon": (80.30, 80.75),
        "zone": "Wet Zone",
        "soil": "Ultisols",
        "ph": 5.0,
        "oc": 2.5,
        "n": 1.55,
        "p": 17,
        "k": 135,
        "sand": 45,
        "clay": 35
    },

    "Kegalle": {
        "lat": (7.05, 7.35),
        "lon": (80.20, 80.55),
        "zone": "Wet Zone",
        "soil": "Ultisols",
        "ph": 5.3,
        "oc": 2.1,
        "n": 1.40,
        "p": 15,
        "k": 130,
        "sand": 48,
        "clay": 32
    }
}


# ================================================================
# 4. RANDOM DISTRICT SELECTION
# ================================================================

district_names = list(districts.keys())

selected_districts = np.random.choice(
    district_names,
    size=N,
    replace=True
)


# ================================================================
# 5. GENERATE RECORDS
# ================================================================

records = []


for i, district in enumerate(selected_districts):

    info = districts[district]


    # ------------------------------------------------------------
    # LATITUDE
    # ------------------------------------------------------------

    latitude = np.random.uniform(
        info["lat"][0],
        info["lat"][1]
    )


    # ------------------------------------------------------------
    # LONGITUDE
    # ------------------------------------------------------------

    longitude = np.random.uniform(
        info["lon"][0],
        info["lon"][1]
    )


    # ------------------------------------------------------------
    # SOIL pH
    # ------------------------------------------------------------

    ph = np.random.normal(
        info["ph"],
        0.35
    )

    ph = np.clip(
        ph,
        4.0,
        8.5
    )


    # ------------------------------------------------------------
    # ORGANIC CARBON
    # ------------------------------------------------------------

    organic_carbon = np.random.normal(
        info["oc"],
        info["oc"] * 0.20
    )

    organic_carbon = np.clip(
        organic_carbon,
        0.30,
        6.00
    )


    # ------------------------------------------------------------
    # NITROGEN
    #
    # Organic carbon influences nitrogen.
    # ------------------------------------------------------------

    nitrogen = (
        info["n"]
        + 0.25 * (organic_carbon - info["oc"])
        + np.random.normal(0, 0.15)
    )

    nitrogen = np.clip(
        nitrogen,
        0.20,
        3.50
    )


    # ------------------------------------------------------------
    # PHOSPHORUS
    #
    # Slight relationship with organic carbon and pH.
    # ------------------------------------------------------------

    phosphorus = (
        info["p"]
        + 2.5 * (organic_carbon - info["oc"])
        - 0.8 * (ph - info["ph"])
        + np.random.normal(0, 4)
    )

    phosphorus = np.clip(
        phosphorus,
        3,
        80
    )


    # ------------------------------------------------------------
    # POTASSIUM
    #
    # Relationship with organic carbon and clay.
    # ------------------------------------------------------------

    potassium = (
        info["k"]
        + 10 * (organic_carbon - info["oc"])
        + np.random.normal(0, 12)
    )

    potassium = np.clip(
        potassium,
        30,
        300
    )


    # ------------------------------------------------------------
    # SOIL TEXTURE
    # ------------------------------------------------------------

    sand = np.random.normal(
        info["sand"],
        6
    )

    clay = np.random.normal(
        info["clay"],
        5
    )


    # ------------------------------------------------------------
    # CLIP SAND AND CLAY
    # ------------------------------------------------------------

    sand = np.clip(
        sand,
        10,
        85
    )

    clay = np.clip(
        clay,
        5,
        60
    )


    # ------------------------------------------------------------
    # SILT
    #
    # IMPORTANT:
    # Sand + Clay + Silt = 100
    # ------------------------------------------------------------

    silt = 100 - sand - clay


    # ------------------------------------------------------------
    # REGENERATE IF SILT IS UNREALISTIC
    # ------------------------------------------------------------

    while silt < 5 or silt > 65:

        sand = np.random.normal(
            info["sand"],
            6
        )

        clay = np.random.normal(
            info["clay"],
            5
        )

        sand = np.clip(
            sand,
            10,
            85
        )

        clay = np.clip(
            clay,
            5,
            60
        )

        silt = 100 - sand - clay


    # ------------------------------------------------------------
    # STORE RECORD
    # ------------------------------------------------------------

    records.append({

        "Sample_ID": f"SL_SOIL_{i+1:05d}",

        "District": district,

        "Latitude": latitude,

        "Longitude": longitude,

        "Agro_Ecological_Zone": info["zone"],

        "Soil_Order_Classification": info["soil"],

        "pH": ph,

        "Nitrogen": nitrogen,

        "Phosphorus": phosphorus,

        "Potassium": potassium,

        "Organic_Carbon": organic_carbon,

        "Sand": sand,

        "Clay": clay,

        "Silt": silt
    })


# ================================================================
# 6. CREATE DATAFRAME
# ================================================================

soil_df = pd.DataFrame(records)


# ================================================================
# 7. ROUND NUMERICAL VALUES
# ================================================================

soil_df["Latitude"] = soil_df["Latitude"].round(6)

soil_df["Longitude"] = soil_df["Longitude"].round(6)

soil_df["pH"] = soil_df["pH"].round(2)

soil_df["Nitrogen"] = soil_df["Nitrogen"].round(3)

soil_df["Phosphorus"] = soil_df["Phosphorus"].round(2)

soil_df["Potassium"] = soil_df["Potassium"].round(2)

soil_df["Organic_Carbon"] = soil_df["Organic_Carbon"].round(2)

soil_df["Sand"] = soil_df["Sand"].round(2)

soil_df["Clay"] = soil_df["Clay"].round(2)

soil_df["Silt"] = soil_df["Silt"].round(2)


# ================================================================
# 8. CHECK DATASET SIZE
# ================================================================

print("\n==============================================")
print("DATASET SIZE")
print("==============================================")

print("Rows:", soil_df.shape[0])

print("Columns:", soil_df.shape[1])


# ================================================================
# 9. DISPLAY FIRST RECORDS
# ================================================================

print("\n==============================================")
print("FIRST 10 RECORDS")
print("==============================================")

display(soil_df.head(10))


# ================================================================
# 10. CHECK MISSING VALUES
# ================================================================

print("\n==============================================")
print("MISSING VALUES")
print("==============================================")

print(
    soil_df.isnull().sum()
)


# ================================================================
# 11. CHECK DUPLICATES
# ================================================================

print("\n==============================================")
print("DUPLICATE RECORDS")
print("==============================================")

print(
    "Duplicates:",
    soil_df.duplicated().sum()
)


# ================================================================
# 12. CHECK DISTRICT DISTRIBUTION
# ================================================================

print("\n==============================================")
print("DISTRICT DISTRIBUTION")
print("==============================================")

print(
    soil_df["District"].value_counts()
)


# ================================================================
# 13. CHECK TEXTURE SUM
# ================================================================

texture_sum = (
    soil_df["Sand"]
    + soil_df["Clay"]
    + soil_df["Silt"]
)

print("\n==============================================")
print("SOIL TEXTURE VALIDATION")
print("==============================================")

print(
    "Minimum:",
    texture_sum.min()
)

print(
    "Maximum:",
    texture_sum.max()
)

print(
    "Mean:",
    texture_sum.mean()
)

print(
    "All approximately 100:",
    np.allclose(
        texture_sum,
        100,
        atol=0.05
    )
)


# ================================================================
# 14. DESCRIPTIVE STATISTICS
# ================================================================

print("\n==============================================")
print("DESCRIPTIVE STATISTICS")
print("==============================================")

display(
    soil_df.describe()
)


# ================================================================
# 15. CHECK SOIL PROPERTY RANGES
# ================================================================

print("\n==============================================")
print("SOIL PROPERTY RANGES")
print("==============================================")

for column in [
    "pH",
    "Nitrogen",
    "Phosphorus",
    "Potassium",
    "Organic_Carbon",
    "Sand",
    "Clay",
    "Silt"
]:

    print(
        f"{column:20s}: "
        f"{soil_df[column].min():.2f} - "
        f"{soil_df[column].max():.2f}"
    )


# ================================================================
# 16. CORRELATION MATRIX
# ================================================================

numeric_columns = [
    "pH",
    "Nitrogen",
    "Phosphorus",
    "Potassium",
    "Organic_Carbon",
    "Sand",
    "Clay",
    "Silt"
]

correlation = soil_df[numeric_columns].corr()

print("\n==============================================")
print("CORRELATION MATRIX")
print("==============================================")

display(
    correlation.round(2)
)


# ================================================================
# 17. PLOT CORRELATION MATRIX
# ================================================================

plt.figure(figsize=(10, 8))

plt.imshow(
    correlation,
    interpolation="nearest"
)

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.colorbar()

plt.title(
    "Soil Properties Correlation Matrix"
)

plt.tight_layout()

plt.show()


# ================================================================
# 18. SAVE CSV
# ================================================================

filename = "soil_data.csv"

soil_df.to_csv(
    filename,
    index=False
)

print("\n==============================================")
print("CSV CREATED")
print("==============================================")

print(
    f"{filename} created successfully!"
)


# ================================================================
# 19. DOWNLOAD CSV
# ================================================================

files.download(filename)

print("\nDownload started.")
```
